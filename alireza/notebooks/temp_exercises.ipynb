```python
def func_from_jomle(m: int, f: str):
    par = "    par{num}: str|None = None,"
    block = "\n".join(list(par.format(num=i) for i in range(1,m)))
    jomle = (
        """def {func}(\n{block}\n):"""
    )
    code = jomle.format(func=f, block=block)
    return code
```

```python
print(func_from_jomle(3, "tabe"))
```
