```python
class LLM_Registry:
    def __init__(
        self, 
        tiers=MODEL_TIERS, 
        provider_order=TIER_PROVIDER_ORDER, 
        cascade_order=TIER_CASCADE_ORDER, 
        default_tier=DEFAULT_TIER,
        test_mode=TEST_MODE,
        test_tier=TEST_TIER,
    ) -> None:
        self._tiers = tiers
        self._provider_order = provider_order
        self._cascade_order = cascade_order
        self._default_tier = default_tier
        self._test_mode = test_mode
        self._test_tier = test_tier
        self._model_cache: dict[tuple[str, str], BaseChatModel] = {} 
        
    def _resolve_tier(self, tier) -> str:
        tier = tier or self._default_tier
        return self._test_tier if self._test_mode else tier
    
    def _providers_for(self, tier):
        return self._provider_order[tier]
    
    def get(self, tier, provider):
        tier = self._resolve_tier(tier)
        provider = provider or self._providers_for(tier)[0]
        key = (tier, provider)
        if key not in self._model_cache:
            self._model_cache[key] = init_chat_model(**self._tiers[tier][provider])
        return self._model_cache[key]
    
    def fallback_models(self, tier) -> list[BaseChatModel]:
        tier = self._resolve_tier(tier)
        providers = [self._providers_for(tier)[1:]]
        return [self.get(tier,p) for p in providers]
```
